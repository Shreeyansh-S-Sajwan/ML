```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

df1 = pd.read_csv('train.csv')
df2 = pd.read_csv('test.csv')

age_median = df1['Age'].median()
fare_median = df1['Fare'].median()

df1['Age'] = df1['Age'].fillna(age_median)
df1['Fare'] = df1['Fare'].fillna(fare_median)
df2['Age'] = df2['Age'].fillna(age_median)
df2['Fare'] = df2['Fare'].fillna(fare_median)

df1['Sex'] = df1['Sex'].map({'male': 0, 'female': 1})
df1 = pd.get_dummies(df1, columns=['Embarked'], drop_first=True)

df2['Sex'] = df2['Sex'].map({'male': 0, 'female': 1})
df2 = pd.get_dummies(df2, columns=['Embarked'], drop_first=True)

features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked_Q', 'Embarked_S']
X_train = df1[features]
y_train = df1['Survived']
X_test = df2[features] 

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model = LogisticRegression(random_state=42)
model.fit(X_train_scaled, y_train)

predictions = model.predict(X_test_scaled)

submission = pd.DataFrame({
    "PassengerId": df2["PassengerId"],
    "Survived": predictions
})
submission.to_csv('submission.csv', index=False)
print("Submission file successfully saved as 'submission.csv'!")
```
